1. import libraries

```python
#import 
%run 00_function.ipynb
# from function import yyplot_k, search_highly_correlated_variables_cv, boruta_cv

import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
from adjustText import adjust_text
from sklearn.linear_model import ARDRegression
from sklearn.model_selection import KFold
from sklearn.metrics import r2_score, mean_absolute_error
from sklearn.preprocessing import StandardScaler
```

2. Setting

```python
#random seed
rseed_cv = 42
rseed_boruta = 1
```

```python
#select dataset, monomer, descriptors
data_set = "1&2" #1&2, 3&4
monomer = "nBA" #St or nBA
descriptors = "mechanism_oriented" #morganFP, RDKit, Mordred, mechanism_oriented
perc = 10 #dataset1&2:St=100, nBA=10, dataset4=50
```

3. Run model

```python
if __name__ == "__main__":
    # file & preprocessing
    try:     
        df_X = pd.read_excel("../data/XY/data_set_{}/data_set_{}_{}_descriptors.xlsx".format(data_set, data_set, descriptors), index_col = 0, sheet_name = monomer) #descriptors select
    except ValueError:
        print(f"Sheet '{monomer}' not found. Loading the first sheet instead.")
        df_X = pd.read_excel("../data/XY/data_set_{}/data_set_{}_{}_descriptors.xlsx".format(data_set, data_set, descriptors), index_col = 0)
        
    df_Y = pd.read_excel("../data/XY/data_set_{}/data_set_{}_SMILES&objective_function.xlsx".format(data_set, data_set), index_col = 0)
    
    
    # X, y 
    df = pd.concat([df_X, df_Y], axis=1)

    # descriptors
    X = df_X.copy()
    y = np.log(df["{}_PDI".format(monomer)]-1) # St_PDI or nBA_PDI
    
    #leverage              
    X = search_highly_correlated_variables(X, 0.8)
    X = boruta(X, y, perc = perc, rseed_boruta = rseed_boruta)
    
    scaler_X = StandardScaler()
    scaler_X.fit(X)
    X_sc = pd.DataFrame(scaler_X.transform(X), index = X.index, columns = X.columns)
     
    n, p = X_sc.shape
    print(n, p)

    # Add 1 to the column to accommodate the intercept in the linear model
    X_with_intercept = np.hstack([np.ones((n, 1)), X_sc.values])  # (n, p+1)


    # H = X (X^T X)^(-1) X^T 
    XtX_inv = np.linalg.inv(X_with_intercept.T @ X_with_intercept)
    H = X_with_intercept @ XtX_inv @ X_with_intercept.T
    leverages = np.diag(H)  # h_i

    leverage_series = pd.Series(leverages, index=X_sc.index, name="leverage")

    print("Leverages (first 5):")
    print(leverage_series.head())
    
    
    
    # ================
    # 4. LOOCV (KFold, n_splits = n) 
    # ================
    fold = len(df.index)  # LOOCV
    kf = KFold(n_splits=fold, shuffle=True, random_state=rseed_cv)
    
    model = ARDRegression(n_iter=1000)
       
    CV_results = pd.DataFrame(index = X.index, columns = ['inner_R2', 'pred_y', 'observed_y', "top10_features", "top10_coefficients"])
    

    for ifold, (tridx, tsidx) in enumerate(kf.split(X, y)):
        try:
            Xtr, Xts = X.iloc[tridx], X.iloc[tsidx]
            ytr, yts = y.iloc[tridx], y.iloc[tsidx]

            Xtr, Xts = search_highly_correlated_variables_cv(Xtr, Xts, 0.8)
            Xtr, Xts = boruta_cv(Xtr, ytr, Xts, perc = perc, rseed_boruta = rseed_boruta) # select the best perc 

            #No standardization when using morganFP               
            scaler_X = StandardScaler()
            scaler_X.fit(Xtr)
            Xtr = pd.DataFrame(scaler_X.transform(Xtr), index = Xtr.index, columns = Xtr.columns)
            Xts = pd.DataFrame(scaler_X.transform(Xts), index = Xts.index, columns = Xts.columns)

            model.fit(Xtr, ytr)
            ytr_pred = model.predict(Xtr)
            yts_pred = model.predict(Xts)

            coefficients = model.coef_
            top10_features = np.argsort(np.abs(coefficients))[::-1][:10].tolist()
            top10_coefficients = [round(num, 2) for num in coefficients[top10_features].tolist()]



            CV_results.loc[Xts.index, "inner_R2"] = r2_score(np.exp(ytr) + 1, np.exp(ytr_pred) + 1)
            CV_results.loc[Xts.index, 'pred_y'] = np.exp(yts_pred) + 1
            CV_results.loc[Xts.index, 'observed_y'] = np.exp(yts) + 1
            CV_results.loc[Xts.index, 'top10_features'] = str(Xtr.columns[top10_features])
            CV_results.loc[Xts.index, 'top10_coefficients'] = str(top10_coefficients)

        except ValueError as e:
            print("An error occurred:", e, "Because the number of descriptors was set to zero by Boruta")    
            sys.exit(1)

    # ================
    # 5. LOOCV Residuals & Standardized Residuals
    # ================
    CV_results['residuals_loo'] = CV_results['observed_y'] - CV_results['pred_y']

    residuals_loo = CV_results['residuals_loo'].astype(float).values
    s = residuals_loo.std(ddof=1)

    CV_results['standardized_residuals'] = residuals_loo / s

    CV_results = CV_results.join(leverage_series)

    print("CV_results (head):")
    print(CV_results.head())
    
    # ================
    # 6. Williams plot 
    # ================
    h_star = 2 * p  / n  # Commonly used leverage cutoffs
    resid_limit = 3      # Standardized residual ±3 is a guideline

    plt.figure(figsize=(6, 5))
    plt.scatter(
        CV_results['leverage'],
        CV_results['standardized_residuals']
    )
    
    if 41 in CV_results.index:
        plt.scatter(
            CV_results.loc[41, 'leverage'],
            CV_results.loc[41, 'standardized_residuals'],
            color="red",
            s=60,
            label="Sample 41"
        )

    for idx in CV_results.index:
        x = CV_results.loc[idx, 'leverage']
        y = CV_results.loc[idx, 'standardized_residuals']
        if idx == 41:
            plt.text(x, y, str(idx), fontsize=8, ha='left', va='bottom', color="red")
        else:
            plt.text(x, y, str(idx), fontsize=8, ha='left', va='bottom', color="steelblue")

    plt.axvline(h_star, linestyle="--", color="gray")
    plt.axhline(resid_limit, linestyle="--", color="orange")
    plt.axhline(-resid_limit, linestyle="--", color="orange")

    plt.xlabel("Leverage (h)")
    plt.ylabel("Standardized residuals")
    plt.title("Williams plot")

    plt.tight_layout()
    plt.savefig(f"williams_plot_{monomer}.jpg", dpi=300)
    plt.show()

    CV_results.to_csv(f"CV_results_with_leverage_{monomer}.csv")
```
